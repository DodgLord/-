Производим чтение файла в список data

```python
data=[]
with open('pr5_2.txt') as F:
  for line in F:
    digit=int(line)
    data.append(digit)
    print(len(data))
print(data[:10])
```

Найдем наименьшее число во всем списке, оканчивающееся на цифру 1

```python
min_end_1 = 10001
for x in data:
  if abs(x) % 10==1 and x<min_end_1:
    min_end_1 = x
```

```python
counter = 0
max_sum = -20001
for i in range(len(data) - 1):
  x, y = data[i], data[i + 1]
  cond1 = abs(x) % 10 == abs(y) % 10
  cond2 = (x % 3 == 0) != (y % 3 == 0)
  cond3 = (x**2 + y**2) <= min_end_1**2
  if cond1 and cond2 and cond3:
    counter += 1
    max_sum = max(max_sum, x + y)

print(f'Количество подходящих пар = {counter}')
print(f'Максимальная сумма = {max_sum}')
```

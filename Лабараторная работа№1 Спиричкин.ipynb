Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: t, z, f1, U.  
Переменная, которая должна быть вычислена на основе независимых: G11.

```python
t=2.8
z=5.0
f1=1.5
U=10.3
print("t =", t, ", type(t) =", type(t))
print("z =", z, ", type(z) =", type(z))
print("f1 =", f1, ", type(f1) =", type(f1))
print("U =", U, ", type(U) =", type(U))
```

Вычисление выражение, печать результатов

Код для вычисления значения выражения G11:

```python
g11 = t * z + z * f1 - (f1 + U) / t
print("g11 =", g11)
print("Type of g11:", type(g11))
```

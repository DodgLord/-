В файле data_v2.txt посчитайте количество символов, расположенных между гласными буквами

Решение

---







```python
f=open('data_v2.txt')
s=f.read().strip()
print(f"Общая длина строки: {len(s)}")
vowels=set("AEIOUYaeiouy")
first_vowel_idx=-1
last_vowel_idx=-1
for i in range(len(s)):
    if s[i] in vowels:
        if first_vowel_idx==-1:
            first_vowel_idx=i
        last_vowel_idx=i
if first_vowel_idx==-1 or first_vowel_idx==last_vowel_idx:
    count=0
else:
    count=0
    for i in range(first_vowel_idx + 1, last_vowel_idx):
        if s[i] not in vowels:
            count+=1
print(f"Количество символов между гласными буквами = {count}")
```

Ответы на контрольные вопросы.

1. Функции read, readline и readlines в Python предназначены для чтения данных из открытого текстового файла. Из сходств они имеют: являются методами файлового объекта, эти функций продвигают внутренний указатель файла вперед, а следующее чтение начнется оттуда, где закончилось предыдущее.
Отличие между ними заключается в объеме данных, который они считывают за один вызов, и формате возвращаемого значения.
2. Что бы прочитать многострочный текстовый фаил это перечислить сам объект файла в цикле for.
3. Главной задачей закрытия файла является освобождение системных ресурсов и сохранение данных на диске.
4. Позиция считывания файла - это числовой адрес внутри файла, который указывает, с какого места будет производиться следующая операция чтения или записи данных. При открытии файла ввода  указатель текущей позиции находится в самом начале файла .

*Решение*

```python
def sum_cubes(n):
    s = 0
    for i in range(1, n + 1):
        s += i*i*i
    return s
```

Демонстрация использования функции

```python
n = 3
print(f'Сумма кубов чисел на отрезке [1; {n}]: {sum_cubes(n)}')

n = 10
print(f'Сумма кубов чисел на отрезке [1; {n}]: {sum_cubes(n)}')

n = 100
print(f'Сумма кубов чисел на отрезке [1; {n}]: {sum_cubes(n)}')
```
